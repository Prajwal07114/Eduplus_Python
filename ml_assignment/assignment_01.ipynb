```python
# 3. Import a CSV file (use any sample dataset) into a DataFrame.
#     Display the number of rows and columns using shape, and print column names.

df = pd.read_csv(r"C:\Users\Prajwal barsagade\Desktop\ML\Pandas-final_output.csv")
```

```python
stud_rec = {
  'Name':["Prajwal","Yash","Harshal","Durgesh","Joe"],
  'Science':[70,80,90,56,98],
  'Maths':[50,60,80,55,69],
  'English':[90,80,70,61,79]
}
df = pd.DataFrame(stud_rec)
print(df)
print(df.head(3))
```

```python
df['Total_Marks'] = df['English'] + df['Maths'] + df['Science']
print(df)

df['Avg_marks'] = df['Total_Marks'] / 3 
print(df)
```

```python
df.loc[4,'Science']=None
print(df)
```

```python
print(df.isnull())
```

```python
df['Science']=df['Science'].fillna(df['Science'].mean())
print(df)
```

```python
# 6. Filter and display all students whose "Average" score is greater than 75.

print(df[df['Avg_marks']>75])
```

```python
# 7. Sort the DataFrame by "Total_Marks" in descending order and show the top 3 performers.

hello = df.sort_values(by='Total_Marks',ascending=False)
print(hello)

print(hello.head(3))
```

```python
# 8. Group students based on their "English" score range (like above or below 70)
grp_data = df.groupby('English')
print(grp_data)
grp_ad = df.groupby('English')['Science'].mean()
print(grp_ad)
```

```python
# 9. Merge two DataFrames:
#     (a) One with "Student_Name" and "Total_Marks"

stud_rec = pd.DataFrame({
  'Name':["Prajwal","Yash","Harshal","Durgesh","Joe"],
  'Score':[70,80,90,56,98],
})

mg = pd.merge(df,stud_rec,on='Name')
print(mg)
```

```python
#     Then create a new column "Overall_Score" = Total_Marks + Sports_Score.

mg['Overall_Score'] = mg['Total_Marks'] + mg['Score']
print(mg)
```

```python
# 10. Export the final DataFrame to a CSV file on your Desktop (use os library to define path)
#     and verify if the file is successfully created using os.path.exists().

mg.to_csv('Pandas-final_output.csv',index=False)
print("Data saved")
```
