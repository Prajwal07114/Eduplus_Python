```python

# Q2: Plot sin(x) and cos(x) on the same graph with different colors, line styles, and labels, and display a legend.

x=np.linspace(0,10,100)
y1=np.sin(x)
y2=np.cos(x)

plt.plot(x,y1,color='red',linestyle='--',marker='*',label='sin(x)')
plt.plot(x,y2,color='blue',linestyle='--',marker='o',label='cos(x)')

plt.title("Same Graph for x&y")
plt.xlabel("X_label")
plt.ylabel("y_label")

plt.grid()
plt.legend()

plt.show()
```

```python
# Q3: Create a bar chart showing the sales of four products: A, B, C, and D with random values.

Products = ['A','B','C','D']
Values = [15,30,45,60]
plt.bar(Products,Values,color='pink')

plt.title("Products Bar_Graph")
plt.xlabel('Products')
plt.ylabel('Values')
```

```python
# Q4: Generate a scatter plot with random x and y values between 0 and 1, and use different colors for points

x = np.random.rand(100)
y = np.random.rand(100)
colors = np.random.rand(100)
size = 540*np.random.rand(100)

plt.scatter(x,y,c=colors,s=size,alpha=0.9,cmap='viridis')
plt.title("Scatter Plot")
plt.xlabel("x-axis")
plt.ylabel('y-axis')

plt.colorbar()
plt.show()
```

```python
# Q5: Plot a histogram for 1000 random numbers generated from a normal distribution.

data = np.random.randn(1000)
plt.hist(data,color='red',bins=37,alpha=0.6,edgecolor='brown',label='Distribution')

plt.title("Histogram")
plt.xlabel("x-label")
plt.ylabel("y-label")
plt.legend()
plt.show()
```

```python
# Q7: Create two subplots in one figure – one showing a sine wave and the other showing a cosine wave.

x = np.linspace(0,15,100)
y1=np.sin(x)
y2=np.cos(x)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.plot(x,y1,'--','o',color='red')
plt.title("Sin(x)")

plt.subplot(1,2,2)
plt.plot(x,y2,'--','*',color='blue')
plt.title("cos(x)")

plt.tight_layout()
plt.show()
```

```python
# Q9: Create a 2x2 grid of subplots showing different plots: line, bar, scatter, and histogram.
subplot=(2,2,1)
fig,axs = plt.subplots(2,2,figsize=(10,8))

x=np.linspace(0,10,100)
axs[0,0].plot(x,np.sin(x),'r')
axs[0,0].set_title("Sine Wave")

#Bar Chart
subplot=(2,2,2)
catogories = ['A','B','C','D']
values = [10,15,7,12]
axs[0,1].bar(catogories,values,color='teal')
axs[0,1].set_title("Bar Chart")

#Histogram
subplot=(2,2,3)
data = np.random.randn(1000)
axs[1,0].hist(data,bins=25,color='purple',edgecolor='black')
axs[1,0].set_title("Histogram")

#Scatter Plot
subplot=(2,2,4)
x=np.random.rand(50)
y=np.random.rand(50)
axs[1,1].scatter(x,y,color='orange')
axs[1,1].set_title("Scatter plot")

plt.suptitle("que-9",fontsize=20)
plt.tight_layout()

plt.savefig("Que_9",dpi=300,bbox_inches='tight')

plt.show()
```

```python

# Q10: Create a line plot for y = e^x for x between 0 and 5, and change the color, marker, and line style.
x=np.linspace(0,5,100)
y=np.exp(x)

plt.plot(x,y,'m--',)
plt.title("Exponential Function")
plt.xlabel("X")
plt.ylabel("exp(X)")

plt.savefig("matplotlib-library-exponential_plot.png",dpi=300,bbox_inches='tight')

plt.show()

```
