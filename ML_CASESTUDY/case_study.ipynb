```python
# Q3: Introduce 10% missing values randomly in Rating and Review columns
# Detect missing values and display the count of missing values per column
# Fill missing ratings with the median rating and missing reviews with "No review provided"
import numpy as np

np.random.seed(56)

rating_mask = np.random.rand(len(df))<0.1
df.loc[rating_mask,'ratings']=np.nan

review_mask = np.random.rand(len(df))<0.1
df.loc[review_mask,'reviews']=np.nan

print(df.isnull().sum())

median_rating = df['ratings'].median()
df.fillna(median_rating,inplace=True)
reviews = df.fillna("No review Provided",inplace=True)
print(df)
```

```python
# Q4: Create a new column 'TotalAmount' by multiplying Quantity and Price
# Add another column 'DiscountedPrice' that applies 10% discount on orders above $100
# Use conditional selection to filter orders where TotalAmount > $150

df['TotalAmount'] = df['quantity']*df['prices']

filtered_order = df[df['TotalAmount']>12000]
print(filtered_order)
```

```python
# Q5: Select only the columns: Product, Category, TotalAmount, and Rating
# Filter and display all orders from the 'Electronics' category with Rating >= 4
electronics = df.loc[
    (df['categories'] == 'Electronics') & (df['ratings'] >= 4),
    ['products', 'categories', 'TotalAmount', 'ratings']
]

print(electronics)
```

```python
# Q6: Use boolean indexing to find all orders where:
# - Quantity > 2 AND Price < 50 OR Rating == 5
# Display the count of such orders

filtered_orders = df[
    ((df['quantity'] > 2) & (df['prices'] < 50)) | (df['ratings'] == 5)
]

print("Number of matching orders:", filtered_orders.shape[0])
```

```python
# Q7: Sort the entire DataFrame by OrderDate (descending) and then by TotalAmount (descending)
# Create a new column 'SalesRank' that ranks products by TotalAmount in descending order

rankod=df.sort_values(by=['order_dates','TotalAmount'],ascending=[False,False])

(rankod['SalesRank'])=rankod['TotalAmount'].rank(ascending=False)
print(rankod[['products', 'TotalAmount', 'SalesRank']].head())
```

```python

# Q8: Find the top 5 best-selling products based on total quantity sold
# Display the product name and total quantity in descending order

top_5_products = (
    df.groupby('products')['quantity']
      .sum()
      .sort_values(ascending=False)
      .head(5)
)

print(top_5_products)
```

```python
# Q9: Group the data by 'Category' and calculate:
# - Total sales (sum of TotalAmount)
# - Average rating
# - Number of orders
# Display the results sorted by total sales

category_summary = (
    df.groupby('categories')
      .agg(
          TotalSales=('TotalAmount', 'sum'),
          AverageRating=('ratings', 'mean'),
          NumberOfOrders=('OrderID', 'count')
      )
      .sort_values(by='TotalSales', ascending=False)
)

print(category_summary)
```

```python

# Q10: Create a pivot table showing average Price for each Category and Rating combination
# Use groupby with multiple columns (Category, Product) and aggregate multiple functions


pivot_table = df.pivot_table(
    values='prices',
    index='categories',
    columns='ratings',
    aggfunc='mean'
)

print(pivot_table)

grouped = df.groupby(['categories', 'products']).agg(
    AvgPrice=('prices', 'mean'),
    TotalSales=('TotalAmount', 'sum'),
    AvgRating=('ratings', 'mean'),
    OrderCount=('products', 'count')
).reset_index()

print(grouped)

```

```python
# Q11: Convert the Quantity and Price columns to NumPy arrays
# Create a 2D array combining these two columns
# Display array properties: shape, dtype, ndim, size
import numpy as np

qty_array = df['quantity'].to_numpy()
price_array = df['prices'].to_numpy()

combined_array = np.column_stack((qty_array, price_array))


print("Combined Array:")
print(combined_array)

print("\nArray Properties:")
print("Shape:", combined_array.shape)
print("Data Type:", combined_array.dtype)
print("Number of Dimensions:", combined_array.ndim)
print("Size:", combined_array.size)
```

```python
# Q12: Create a 1D NumPy array of 100 random prices between 10 and 500
# Calculate mean, median, standard deviation, min, and max using NumPy functions
# Reshape this array into a 10x10 matrix

prices = np.random.uniform(10,500,100)

mean = np.mean(prices)
median = np.median(prices)
std = np.std(prices)
min = np.min(prices)
max = np.max(prices)


print("mean :",mean)
print("max  :",max)
print("min  :",min)
print("median :",median)
print("std :",std)



prices_matrix = prices.reshape(10, 10)

print("\n10x10 Matrix:\n", prices_matrix)
```
